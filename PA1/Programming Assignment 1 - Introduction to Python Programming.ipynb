## Programming Problem A - Word Rotation Problem

```python
def rotate_word(text):
    return(text[1:] + text[0])
```

```python
rotate_word("python")
```

```python
rotate_word("logic")
```

```python
rotate_word("Code")
```

```python
rotate_word("A")
```

## Programming Problem B - Username Builder Problem

```python
def make_username(first_name, last_name):
    fname = first_name.lower()
    surname = last_name.lower()
    return(fname.replace(" ","") + "." + surname.replace(" ",""))
```

```python
make_username("Ada", "Lovelace")
```

```python
make_username("Alan", "Turing")
```

```python
make_username("Ana Maria", "De Leon")
```

## Programming Problem C - Bookend Swap Problem

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last, *middle, first]
```

```python
swap_bookends([1, 2, 3, 4, 5, 6])
```

```python
swap_bookends(["red", "green", "blue"])
```

```python
swap_bookends([8, 3])
```
