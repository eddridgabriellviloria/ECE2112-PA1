## Programming Problem A - Reproducible Normalization Problem

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X_Normalized = (X-X.mean())/(X.std())
X_Normalized
```

## Programming Problem B - Cubes Divisible By 4 Problem

```python
A = np.arange(1,101,1)
A
```

```python
C = A.reshape(10,10)
C = C*C*C
C
```

```python
div_by_4 = C[C%4==0]
div_by_4
```

```python
np.save("div_by_4",div_by_4)
```

## Programming Problem C - Above-Mean Squares Problem

```python
S = np.arange(1,37,1)
S = S.reshape(6,6)
S = S*S
S
```

```python
S_mean = S.mean()
S_mean
```

```python
above_mean = S[S>S_mean]
above_mean
```
